# checking if the API key available

```python
import requests
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
api_key= os.getenv("OPENAI_API_KEY")

if not api_key:
    print("No api key found. Please set the OPENAI_API_KEY environment variable.")
elif not api_key.startswith("sk-proj-"):
    print("api key is not valid. Please set the OPENAI_API_KEY environment variable to a valid key.")
else:
    print("API key is found and looks good so far")
```

# HTTPS ENDPOINTS
- setting headers and payload


```python
headers = {"Authorization": f"Bearer {api_key}", "content-type": "application/json"} #here we are setting the headers for the request.
payload = {
    "model" : "gpt-5-nano", 
    "input": [    #messages is outdated parameter,new->"input"
        {"role": "user", "content": "tell me a fun fact!"}
    ]
}

payload
```

```python

import json

response = requests.post(
    "https://api.openai.com/v1/responses", #course he was using "https://api.openai.com/v1/chat/completions" but it was giving me 404 error so I changed it to "https://api.openai.com/v1/response"
    headers=headers,
    json=payload
)


print(json.dumps(response.json(), indent=2))
response.status_code
```

# Instead of writing API HTTP end points
- We can use openai package provided by openai 
- package name: Python Client Library
- Instead of deepd diving into json snippets
- We can use python client library, for more readable outputs

```python
from openai import OpenAI
openai = OpenAI()

response= openai.chat.completions.create(model= "gpt-5-nano",
                                         messages=[{"role": "user", "content": "tell me a fun fact!"}])
response.choices[0].message.content
```
