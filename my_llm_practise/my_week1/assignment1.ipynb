Declaring the model and creating OpenAI instances


```python
MODEL = "gpt-5-nano"
openai = OpenAI()
```

USER AND SYSTEM PROMPTS


```python
system_prompt = """
    You are a patient teacher. Explain things simply, with one everyday example.
"""


def user_prompt(concept):
    return f"""Explain {concept} under 100 words."""


print(user_prompt("Tokens"))
```

Creating function **patient_teacher(concept)** REQUESTING OPENAI api.


```python
def patient_teacher(concept):
    display(Markdown(f"# {concept.upper()}"))
    response = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt(concept)},
        ],
        # stream=True,
        # stream_options={"include_usage": True},
    )

    result = response.choices[0].message.content

    display(Markdown(f"### {result}"))

    display(Markdown((f"### Tokens used for Input: {response.usage.prompt_tokens}")))

    display(
        Markdown((f"### Tokens used for Output: {response.usage.completion_tokens}"))
    )
    display(
        Markdown(
            f"### Total tokens used for this output: {response.usage.completion_tokens} + {response.usage.prompt_tokens} = {response.usage.prompt_tokens} + {response.usage.completion_tokens}"
        )
    )
```

### WE ARE PRINTING LIKE AI GENERATED OUTPUT ANIMATION

- Using stream parameter
  - stream = True while requesting an API call.
  - Then breaking the output in chunks
    - then printing the output, using for loop.


```python
def patient_teacher1(concept):
    display(Markdown((f"# {concept.upper()}")))
    stream = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt(concept)},
        ],
        stream=True,
    )

    result = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        result += chunk.choices[0].delta.content or ""
        update_display(Markdown(result), display_id=display_handle.display_id)
```

```python
patient_teacher1("Tokens")
patient_teacher1("Context Window")
patient_teacher1("Frontier Model")
```
